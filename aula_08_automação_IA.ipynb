```python
# @title
# Script Python para verificar a temperatura do sistema

# Simula a leitura da temperatura (você pode substituir isso por uma leitura real)
temperatura = float(input("Digite a temperatura atual do sistema: "))

# Verifica se a temperatura é maior que 30 graus
if temperatura > 30:
    print("ALERTA: Superaquecimento detectado! Temperatura acima de " + str(temperatura) + "°C.")
else:
    print("Temperatura do sistema está normal.")
```
